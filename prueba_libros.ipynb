```python
import requests
from bs4 import BeautifulSoup
import csv
import time

libros = []
numero_pagina = 1

while True:  # bucle que se repite indefinidamente... hasta que lo paremos nosotros
    url = f"https://books.toscrape.com/catalogue/page-{numero_pagina}.html"
    print(f"Descargando página {numero_pagina}...")

    try:
        respuesta = requests.get(url, timeout=10)
    except requests.exceptions.RequestException as error:
        print(f"Error de conexión: {error}")
        break  # rompe el bucle, no tiene sentido seguir si no hay internet

    # Si la página no existe, el servidor devuelve 404 -> hemos pasado del final
    if respuesta.status_code == 404:
        print("No hay más páginas. Terminado.")
        break

    respuesta.encoding = "utf-8"
    sopa = BeautifulSoup(respuesta.text, "html.parser")

    todos_los_h3 = sopa.find_all("h3")
    todos_los_precios = sopa.find_all("p", class_="price_color")

    for h3, precio in zip(todos_los_h3, todos_los_precios):
        enlace = h3.find("a")
        titulo = enlace["title"]
        precio_texto = precio.text
        libros.append({"titulo": titulo, "precio": precio_texto})

    # Comprobamos si existe el botón "next" -> si NO existe, es la última página
    boton_siguiente = sopa.find("li", class_="next")
    if boton_siguiente is None:
        print("Última página alcanzada (no hay botón 'next').")
        break

    numero_pagina += 1  # pasamos a la siguiente página
    time.sleep(1)

with open("libros.csv", "w", newline="", encoding="utf-8-sig") as archivo:
    escritor = csv.DictWriter(archivo, fieldnames=["titulo", "precio"])
    escritor.writeheader()
    escritor.writerows(libros)

print(f"Se guardaron {len(libros)} libros en libros.csv")
```
